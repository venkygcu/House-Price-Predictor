# California Housing: Feature Scaling and Model Comparison

Task 2 extends the baseline by comparing Linear Regression, Ridge Regression, and a depth-limited Decision Tree with leakage-safe preprocessing. The target is in $100,000s of 1990 USD.

## 1. Load and inspect the data

The scikit-learn California Housing dataset contains eight numeric features and the median house value target. Running the workflow creates summary statistics, missing-value counts, distributions, and target correlations in `reports/`.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import Markdown, display

ROOT = Path.cwd()
if not (ROOT / 'house_price_predictor.py').exists():
    ROOT = ROOT.parent
from house_price_predictor import main

# Run the complete reproducible workflow and refresh all outputs.
main()
```

## 2. Feature scaling and train/test split

An 80/20 split uses `random_state=42`. Each candidate pipeline performs median imputation, then `StandardScaler`, then model fitting. Both preprocessing steps are fitted on training rows only, preventing test data leakage. Scaling is useful for linear and Ridge models; using a shared pipeline keeps preprocessing consistent for comparison.

## 3. Train and compare models

Candidates are Linear Regression, Ridge (`alpha=1.0`), and Decision Tree (`max_depth=5`). MAE, RMSE, and R? are measured on the same held-out rows. The lowest RMSE selects the final model; lower RMSE indicates lower error and penalizes large errors more strongly.

```python
comparison = pd.read_csv(ROOT / 'reports' / 'model_comparison.csv')
display(comparison)
metrics = json.loads((ROOT / 'reports' / 'metrics.json').read_text(encoding='utf-8'))
display(Markdown(f"**Selected model:** {metrics['selected_model']} (lowest held-out RMSE)."))
display(Markdown(f"**Selected model scores:** MAE = {metrics['mae_100k_usd']:.4f}, RMSE = {metrics['rmse_100k_usd']:.4f} in $100,000s; R? = {metrics['r2']:.4f}."))
```
